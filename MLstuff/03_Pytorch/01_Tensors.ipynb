## Making Tensors

```python
# empty
a=torch.empty(2, 3)

# shows whats in that memory location, not the actual values
```

```python
type(a)
```

```python
#Zeroes or ones
torch.zeros(2,3)
```

```python
#seed+rand
torch.manual_seed(0)
torch.rand(2,3)
```

```python
#My tensor
print(torch.tensor([[1, 2,4 ], [3, 4,5]]))
```

```python
#More ways to create tensors
print(torch.arange(0, 10, step=2)) # 0 to 10 with step of 2

#linspace
print(torch.linspace(0, 10, steps=5)) # 5 points between 0 and 1

print(torch.eye(3))  # identity matrix

print(torch.full((2, 3), 7))  # 2x3 matrix of sevens
```
